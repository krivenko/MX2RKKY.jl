# Performance benchmarks

```julia
using Pkg; Pkg.update()

using MX2RKKY.montecarlo
using MX2RKKY.lattice

using OffsetArrays: OffsetArray, Origin
using Random: MersenneTwister
using BenchmarkTools: @btime, @benchmark
```

## `xy_energy_contrib!()`

```julia
using MX2RKKY.montecarlo: xy_energy_contrib!

L = 27
J = 1.0
conf = Configuration(L, L, -1)
pair_int_energy = OffsetArray(zeros(Float64, L, L), Origin(0))

@benchmark xy_energy_contrib!($conf, (0, 0), $pair_int_energy, J=$J)
```

## `rkky_energy_contrib!()`

```julia
using MX2RKKY.montecarlo: rkky_energy_contrib!, make_rkky_int_array, make_rkky_angle_array

L = 27
conf = Configuration(L, L, -1)
pair_int_energy = OffsetArray(zeros(Float64, L, L), Origin(0))

V = 0.001
λ = 0.1
Varr = make_rkky_int_array(L, L, V, λ)
ϕarr = make_rkky_angle_array(L, L)

i = (0, 0)
@benchmark rkky_energy_contrib!($conf, $i, $pair_int_energy, Val(0), V=$Varr, ϕ=$ϕarr)
```
